# **Import libraries**

```python
import tkinter as tk
from tkinter import ttk
import joblib
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

# **Loading the model**

```python
model = joblib.load('FinalModel.pkl')
```

# **Graphics using tk**

## **The root, frames,calculations and griding the elements**

```python
root = tk.Tk()
root.title("HouseValues in California")

width = 500
height = 600

sc_width = root.winfo_screenwidth()
sc_height = root.winfo_screenheight()

x = (sc_width - width) // 2
y = (sc_height - height) // 2

root.geometry(f"{width}x{height}+{x}+{y}")
root.resizable(False, False)
root.minsize(500,600)
root.config(bg = '#9cb2ba')
```

```python
f1 = tk.Frame(root,bg='#9cb2ba')
fexit = tk.Frame(root,bg='#9cb2ba')
```

```python
def frame_exit():
    fexit.tkraise()

    l1 = tk.Label(fexit,text='Are you sure that you want to exit?',font=('Franklin Gothic Medium',22),bg='#9cb2ba')
    
    def exit_tk():
        root.destroy()

    b1 = tk.Button(fexit,text='Yes',font=('Franklin Gothic Medium',22),bg='#adaeb3',command=exit_tk)
    b2 = tk.Button(fexit,text='No',font=('Franklin Gothic Medium',22),bg='#adaeb3',command=frame_1)    

    l1.grid(row=0,column=0,columnspan=2,padx=5,pady=5,sticky="NEWS")
    b1.grid(row=1,column=0,padx=5,pady=5,sticky="NEWS")
    b2.grid(row=1,column=1,padx=5,pady=5,sticky="NEWS")

    for i in range(2):
        for j in range(2):
            fexit.grid_rowconfigure(i,weight=1)
            fexit.grid_columnconfigure(j,weight=1)
```

```python
def frame_1():
    f1.tkraise()
    
    l1 = tk.Label(f1,text='Find out how much your house costs!',font=('Franklin Gothic Medium',20),bg='#9cb2ba')
    l2 = tk.Label(f1,text='Longitude:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l3 = tk.Label(f1,text='Latitude:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    
    l4 = tk.Label(f1,text='Housing Median Age:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l5 = tk.Label(f1,text='Total Rooms:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    
    l6 = tk.Label(f1,text='Total Bedrooms:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l7 = tk.Label(f1,text='Population:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l8 = tk.Label(f1,text='Households:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    
    l9 = tk.Label(f1,text='Median Income:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l10 = tk.Label(f1,text='Ocean Proximity:',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    l11 = tk.Label(f1,text='',font=('Franklin Gothic Medium',16),bg='#9cb2ba')
    
    e1 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e2 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e3 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e4 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    
    e5 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e6 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e7 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    e8 = tk.Entry(f1,font=('Franklin Gothic Medium',16))
    
    combo_box = ttk.Combobox(f1,values=('NEAR BAY','INLAND','<1H OCEAN','NEAR OCEAN','ISLAND','-'),state='readonly')
    combo_box.set('Please choose an option')
    
    def calculate():    
        try:
            longitude = e1.get()
            latitude = e2.get()
            houseAge = e3.get()
            rooms = e4.get()
            bedrooms = e5.get()
            population = e6.get()
            households = e7.get()
            income_raw = e8.get()
            if income_raw.isdigit():
                income = str(float(income_raw) / 10000)
            elif income_raw == '-':
                income = '-'
            else:
                raise ValueError
            proximity = combo_box.get()
            data_list = [longitude,latitude,houseAge,rooms,bedrooms,population,households,income,proximity]                
            t = 0
            finalList = []
            for i in data_list:
                if i == '-':
                    finalList.append(np.nan)
                    t += 1
                else:
                    finalList.append(i)
            if t >= 7:
                raise IndexError
            else:
                if bedrooms != '-' and rooms != '-' and bedrooms.isdigit() and rooms.isdigit():
                    if int(bedrooms.strip()) >= int(rooms.strip()):
                        raise AttributeError
                if households != '-' and population != '-' and households.isdigit() and population.isdigit():
                    if int(households.strip()) > int(population.strip()):
                        raise AttributeError
                l11.config(text='Calculating ...')
                data = pd.DataFrame([finalList])
                data.columns = ['longitude','latitude','housing_median_age', 'total_rooms', 'total_bedrooms','population','households','median_income','ocean_proximity']

                result = model.predict(data)
                finalResult = result.astype(int)
                def update_label():
                    l11.config(text=f'Your house costs ${finalResult[0]}.')
                root.after(3000,update_label)
        except IndexError:
            l11.config(text="You must enter at least 3 features!")
        except AttributeError:
            l11.config(text='Wrong input for Total Bedrooms or Households!')
        except ValueError:
            l11.config(text="Please fill all the blanks correctly and try again!\nAnd if you don't know any of the values enter (-)")
    b1 = tk.Button(f1,text='Calculate',font=('Franklin Gothic Medium',16),bg='#adaeb3',command=calculate)
    b2 = tk.Button(f1,text='Exit',font=('Franklin Gothic Medium',16),bg='#adaeb3',command=frame_exit)
    
    l1.grid(row=0,column=0,columnspan=2,padx=5,pady=5,sticky="NEWS")
    l2.grid(row=1,column=0,padx=5,pady=5,sticky="NEWS")
    l3.grid(row=2,column=0,padx=5,pady=5,sticky="NEWS")
    
    l4.grid(row=3,column=0,padx=5,pady=5,sticky="NEWS")
    l5.grid(row=4,column=0,padx=5,pady=5,sticky="NEWS")
    l6.grid(row=5,column=0,padx=5,pady=5,sticky="NEWS")
    
    l7.grid(row=6,column=0,padx=5,pady=5,sticky="NEWS")
    l8.grid(row=7,column=0,padx=5,pady=5,sticky="NEWS")
    l9.grid(row=8,column=0,padx=5,pady=5,sticky="NEWS")
    
    l10.grid(row=9,column=0,padx=5,pady=5,sticky="NEWS")
    l11.grid(row=11,column=0,columnspan=2,rowspan=2,padx=5,pady=5,sticky="NEWS")
    
    e1.grid(row=1,column=1,padx=5,pady=5,sticky="NEWS")
    e2.grid(row=2,column=1,padx=5,pady=5,sticky="NEWS")
    e3.grid(row=3,column=1,padx=5,pady=5,sticky="NEWS")
    e4.grid(row=4,column=1,padx=5,pady=5,sticky="NEWS")
    
    e5.grid(row=5,column=1,padx=5,pady=5,sticky="NEWS")
    e6.grid(row=6,column=1,padx=5,pady=5,sticky="NEWS")
    e7.grid(row=7,column=1,padx=5,pady=5,sticky="NEWS")
    e8.grid(row=8,column=1,padx=5,pady=5,sticky="NEWS")
    
    combo_box.grid(row=9,column=1,padx=5,pady=5,sticky="NEWS")
    
    b1.grid(row=10,column=0,padx=5,pady=5,sticky="NEWS")
    b2.grid(row=10,column=1,padx=5,pady=5,sticky="NEWS")
    
    for i in range(13):
        for j in range(2):
            f1.grid_rowconfigure(i,weight=1)
            f1.grid_columnconfigure(j,weight=1)
```

## **Griding the frames and the root**

```python
frame_1()
f1.grid(row=0,column=0,sticky="NEWS")
fexit.grid(row=0,column=0,sticky="NEWS")
root.grid_rowconfigure(0,weight=1)
root.grid_columnconfigure(0,weight=1)
```
